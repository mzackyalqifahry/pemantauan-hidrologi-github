```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000
N = 200
n = np.arange(N)
Ts = 1 / fs
f1 = 50
omega = 2 * np.pi * f1

# operation Amplify signal = x2, advance signal = +10, delay = -10, and mirrored = -1
y = 5 * np.sin(omega * n * Ts) + 4 * np.cos(np.pi * n * Ts)
y_amp = 2 * y
advance_samples = 10
y_adv = np.roll(y, -advance_samples)
delay_samples = 10
y_del = np.roll(y, delay_samples)
y_mir = y[::-1]

# For colouring
def smooth_stem(n, y, color, label):
    markerline, stemlines, baseline = plt.stem(n, y, basefmt=" ", label=label)
    plt.setp(stemlines, color=color, linewidth=0.8, alpha=0.7)
    plt.setp(markerline, color=color, markersize=3, alpha=0.8)

# Plot
plt.figure(figsize=(12, 10))

# a) For Amplified Signal
plt.subplot(4, 1, 1)
smooth_stem(n, y, 'k', 'Original')  # black
smooth_stem(n, y_amp, 'r', 'Amplified (×2)')
plt.title('(a) Amplified Signal')
plt.xlabel('n')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True, alpha=0.3)

# b) For Advanced Signal
plt.subplot(4, 1, 2)
smooth_stem(n, y, 'k', 'Original')  # black
smooth_stem(n, y_adv, 'r', 'Advanced (−10 samples)')
plt.title('(b) Advanced Signal')
plt.xlabel('n')
plt.legend()
plt.grid(True, alpha=0.3)

# c) Delayed Signal
plt.subplot(4, 1, 3)
smooth_stem(n, y, 'k', 'Original')  # black
smooth_stem(n, y_del, 'r', 'Delayed (+10 samples)')
plt.title('(c) Delayed Signal')
plt.xlabel('n')
plt.legend()
plt.grid(True, alpha=0.3)

# d) Mirrored Signal
plt.subplot(4, 1, 4)
smooth_stem(n, y, 'k', 'Original')  # black
smooth_stem(n, y_mir, 'r', 'Mirrored (y[-n])')
plt.title('(d) Mirrored Signal')
plt.xlabel('n')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

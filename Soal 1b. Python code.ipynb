```python
from sympy import symbols, Eq, solve
x, y, z = symbols('x y z')

# define equations
eq1 = Eq(x + y + 2*z, 8)
eq2 = Eq(-x - 2*y + 3*z, 1)
eq3 = Eq(3*x - 7*y + 4*z, 10)

# solve the system of equations
solution = solve((eq1, eq2, eq3), (x, y, z))

# result
print("The original coordinates (x, y, z):", solution)
```

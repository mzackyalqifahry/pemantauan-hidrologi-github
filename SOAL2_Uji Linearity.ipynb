```python
import numpy as np

# Definisi sistem
def system(x):
    return x + 5
x1 = np.array([1, 2, 3])
x2 = np.array([4, 5, 6])

# Skalar
a = 2
b = 3

# Output
y1 = system(x1)
y2 = system(x2)

# Uji Prinsip Superposisi
lhs = system(a * x1 + b * x2)    # kombinasi input
rhs = a * y1 + b * y2            # Kombinasi output

# Print result
print("LHS (system(a*x1 + b*x2)):", lhs)
print("RHS (a*y1 + b*y2):", rhs)

# Print linearity
if np.allclose(lhs, rhs):
    print(" Sistem linear.")
else:
    print("tidak linear.")
```

```python
import numpy as np

# Definisi sistem
def system(x):
    x_shifted = np.roll(x, 2)   # Geser 2 smple ke kanan
    x_shifted[:2] = 0
    return 4*x - 2*x_shifted

x1 = np.array([1, 2, 3, 4, 5])
x2 = np.array([4, 5, 6, 7, 8])

# Skalar
a = 2
b = 3

# Output
y1 = system(x1)
y2 = system(x2)

# Uji Prinsip Superposisi
lhs = system(a * x1 + b * x2)    # kombinasi input
rhs = a * y1 + b * y2            # kombinasi output

# Print results
print("LHS (system(a*x1 + b*x2)):", lhs)
print("RHS (a*y1 + b*y2):", rhs)

# Print linearity
if np.allclose(lhs, rhs):
    print("Sistem linear.")
else:
    print("Sistem tidak linear.")
```

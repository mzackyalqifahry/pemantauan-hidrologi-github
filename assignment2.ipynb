```python
import numpy as np
import matplotlib.pyplot as plt

def generateSineWave(fs, f):
    t = np.arange(0, 1, 1/fs)
    x = 3 * np.sin(2 * np.pi * f * t)
    return x

def DFT(x):
    N = len(x)
    X = np.zeros(N, dtype=complex)
    for k in range(N):
        for n in range(N):
            angle = (2 * np.pi * k * n) / N
            X[k] += x[n] * (np.cos(angle) - 1j * np.sin(angle))
    return X


fs = 100
f = 10

x = generateSineWave(fs, f)
X = DFT(x)

N = len(X)
freq = np.arange(N) * fs / N

# DFT tanpa normalisasi
plt.figure(figsize=(8, 6))
plt.stem(freq, np.abs(X), basefmt="-b")
plt.title('DFT (Tanpa Normalisasi)')
plt.xlabel('(Hz)')
plt.ylabel('Amplitude')
plt.grid(True)

# DFT dengan normalisasi, supaya amplitude nya 3
X = X / (N / 2)

plt.figure(figsize=(8, 6))
plt.stem(freq, np.abs(X), basefmt="-b")
plt.title('DFT (Dengan Normalisasi Amplitudo)')
plt.xlabel('(Hz)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```
